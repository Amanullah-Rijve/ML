```python
import pandas as pd 
import numpy as np 

data = {
    "name": ["Karim", "Rahim", "Salim", "Jamal", "Nabil", "Rafi", "Siam", "Tuhin"],
    "age": [22, 25, 21, 28, 23, 26, 24, 27],
    "department": ["CSE", "SWE", "CSE", "EEE", "SWE", "EEE", "CSE", "SWE"],
    "cgpa": [3.5, 3.8, 2.9, 3.2, 3.9, 3.1, 3.6, 3.7],
    "salary": [45000, 55000, 35000, 50000, 60000, 48000, 52000, 58000]
}
df = pd.DataFrame(data)

# ১। Department অনুযায়ী average CGPA বের করো (groupby)
avgcg = df.groupby("department")["cgpa"].mean()
print(f"Average salary:\n {avgcg}")
```

```python
# ২। Salary অনুযায়ী descending sort করো, top 3 দেখাও
salary_desc = df.sort_values(by="salary",ascending=False).head(3)
print(f"top 3 decending salary: \n {salary_desc}")
```

```python
# ৩। CGPA 3.5 এর বেশি যাদের filter করো
cgpa_check = df[df["cgpa"] > 3.5]
print(f"3.5 upper cgpa: \n {cgpa_check}")
```

```python
# ৪। "performance" column যুক্ত করো:
#     cgpa >= 3.7 → "Excellent"
#     cgpa >= 3.3 → "Good"
#     বাকি → "Average"
def get_performance(cgpa):
    if cgpa>=3.7:
        return "Excellent"
    elif cgpa >=3.3:
        return "Good"
    else: return "Average"

df["performance"]= df["cgpa"].apply(get_performance)
print(df[["name","cgpa","performance"]])
```

```python
# ৫। NumPy দিয়ে salary এর mean, std, max, min বের করো
salary_Arr = np.array(df["salary"])
print(f"Mean: {np.mean(salary_Arr)}")
print(f"Max: {np.max(salary_Arr)}")
print(f"std: {np.std(salary_Arr)}")
print(f"min: {np.min(salary_Arr)}")
```

```python
# ৬। Department অনুযায়ী total salary বের করো (groupby + sum)
total_saalry = df.groupby(df["department"])["salary"].sum()
print("Total salary: \n",total_saalry)
```
