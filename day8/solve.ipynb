```python
import pandas as pd 

data = {
    "name": ["Karim", "Rahim", "Salim", "Jamal", "Nabil"],
    "age": [20, 22, 21, 23, 20],
    "math": [85, 92, 78, 60, 95],
    "english": [70, 88, 65, 75, 90],
    "science": [80, 85, 72, 68, 88]
}
```

```python
df = pd.DataFrame(data)
print('shape: \n', df.shape)
print('col: \n', df.columns)
print('head: \n', df.head())
```

```python
df['average']= (df['math']+ df['english']+df['science'])/3
print('Average added: \n', df)
```

```python
# filter more than 80% marks
filtered = df[df['average']>80]
print('average >80 : \n', filtered)
```

```python
# add grade column
def get_grade(avg):
    if avg >=90:
        return 'A'
    elif avg >= 80:
        return 'B'
    elif avg >=70:
        return 'C'
    else: return 'F'
    
df['grade']=df['average'].apply(get_grade)
print('with grade:\n', df)
```

```python
# save - amr data gula akta csv file e convert korlam
df.to_csv('student.csv',index=False)
print('Csv saved!')
```
