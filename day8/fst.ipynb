```python
# import pandas
import pandas as pd 
#create a list of data
data = [1,2,3,4,5]
#print(data)
#print(type(data))
# create series (1d) of that data
s = pd.Series(data)
print(s)
print(type(s))
#create dataframe (2d) from series
df = pd.DataFrame(s)
print(df)
print(type(df))
''' 
Easy way to remember
Series = one-dimensional
Has Index + Values
DataFrame = two-dimensional
Has Row Index + Column Names + Values
'''
# give name to the col
df = pd.DataFrame(s,columns=['number'])
print(df)
```
