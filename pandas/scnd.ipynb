```python
import pandas as pd

# use pandas with real time data set
# EDA - Exploratory Data Analysis

# online data set
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
# step by step EDA
# step1 - Data dekho
print(df.shape) # see row and column - (891, 12) 891 -row 12-col
print(df.head()) # see first 5 rows
print(df.columns) # see column names
print(df.info()) # give data types and missing inf
print(df.describe()) # stastical summery
```

```python
# step 2 - Missing Data
# missing data of every colum - (how much)
print(df.isnull().sum())

# fix that missing data
df["Age"].fillna(df["Age"].mean(), inplace=True) # fill with mean - avg age# inplace true - modify exsisting data and false return new data
df.drop(columns=["Cabin"],inplace=True) # most missing col remove
df.dropna(inplace=True) # missing row remove
```

```python
# step 3 - groupby()

# survival rate by Sex
print(df.groupby("Sex")["Survived"].mean())
# avg age by class
print(df.groupby("Pclass")["Age"].mean())
```

```python
# step 4 - sort_values()

#  sort by Age
df.sort_values("Age",ascending=False).head(10)
# multiple column sorting
df.sort_values(["Pclass","Age"]).head()
```

```python
# step 5 - value_counts

print(df["Sex"].value_counts()) # male/female count
print(df["Pclass"].value_counts()) # class waise count
print(df["Survived"].value_counts()) # how much Survived
```

```python
# step 6 - new features Engineering 

# make Family Size
df1 = df["FamilySize"] = df["SibSp"] + df["Parch"] +1
print(df1)
# check if Alone
df2 = df["IsAlone"] = (df["FamilySize"]==1).astype(int)
print(df2)
```
