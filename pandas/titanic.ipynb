```python
import pandas as pd 

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print(df.shape)
print(df.head())
print(df.isnull().sum())
```

```python
# filling missing age
df["Age"].fillna(df["Age"].mean(),inplace=True)
# remove cabin column
df.drop(columns=["Cabin"],inplace=True)
# remove missing row
df.dropna(inplace=True)

print(df.isnull().sum())
print(df.shape)
```

```python
# Male female survibal rate
print("Survival By Gender: ")
print(df.groupby("Sex")["Survived"].mean())
# average fare by class
print("\n Average Fare By Class: ")
print(df.groupby("Pclass")["Fare"].mean())
# survibal rate by class
print("\n Survival By class: ")
print(df.groupby("Pclass")["Survived"].mean())
```

```python
# desencding count by age
print(df.sort_values("Age",ascending=False).head(10))
# multiple column sorting
print(df.sort_values(["Pclass","Age"]).head())
```

```python
# make Family Size
df1 = df["FamilySize"] = df["SibSp"] + df["Parch"] +1
print(df1)
# check if Alone
df2 = df["IsAlone"] = (df["FamilySize"]==1).astype(int)
print(df2)
```

```python
# save data inm csv
save= df.to_csv('cleaned_data.csv', index=False)
print(f"Data Saved {save}")
```
