```python
import numpy as np 

# AI model response time
normal_responses = [2.1,2.3,1.9,2.5, 2.2, 2.0, 2.4]

# calculate
mean = np.mean(normal_responses)
median = np.median(normal_responses)
std = np.std(normal_responses)

print(f"Mean:\n {mean:.2f}")
print(f"Median:\n {median:.2f}")
print(f"Std Dev: \n {std:.2f}")
```

```python
# Suspicious Response
suspicious = 0.001
print(F"\n Normal Range: {mean-2*std:.2f} to {mean+2*std:.2f}")
print(f"suspicious response: \n{suspicious}")
print(f"Is attack? {suspicious <mean-2*std:.2f}")
```

```python
# multiple suspicious req
requests = [2.1, 2.3, 0.001, 2.2, 0.002, 
            2.4, 1.9, 0.001, 2.5, 2.1]

print("\n--- Attack Detection ---")
for i,req in enumerate(requests):
    is_attack = req<mean -2*std
    status = "Attack! "if is_attack else"Normal"
    print(f"Request {i+1}: {req}s {status}")
```
